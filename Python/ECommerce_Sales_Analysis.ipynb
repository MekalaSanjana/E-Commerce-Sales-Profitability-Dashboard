```python
df_clean = df.copy()

print("Working copy created successfully!")
```

```python
before = len(df_clean)

df_clean = df_clean.drop_duplicates()

after = len(df_clean)

print("Rows before:", before)
print("Rows after:", after)
print("Duplicates removed:", before - after)
```

```python
df_clean["Month"] = df_clean["Month"].astype(str).str.strip()

print("Month column prepared successfully!")
print(df_clean["Month"].head())
```

```python
df_clean.info()
```

```python
print(df_clean[["Unit Price", "Units Sold", "Discount (%)", "Distance"]].describe())
```

```python
df_clean["Gross_Sales"] = df_clean["Unit Price"] * df_clean["Units Sold"]

print("Gross Sales column created!")
```

```python
# Convert important numeric columns
df_clean["Unit Price"] = pd.to_numeric(
    df_clean["Unit Price"], errors="coerce"
)

df_clean["Units Sold"] = pd.to_numeric(
    df_clean["Units Sold"], errors="coerce"
)

df_clean["Discount (%)"] = pd.to_numeric(
    df_clean["Discount (%)"], errors="coerce"
)

# Recalculate Gross Sales
df_clean["Gross_Sales"] = (
    df_clean["Unit Price"] * df_clean["Units Sold"]
)

# Calculate Discount Amount
df_clean["Discount_Amount"] = (
    df_clean["Gross_Sales"] * df_clean["Discount (%)"] / 100
)

print("Numeric columns converted successfully!")
print("Discount Amount calculated successfully!")
```

```python
print(df_clean[[
    "Unit Price",
    "Units Sold",
    "Discount (%)",
    "Gross_Sales",
    "Discount_Amount"
]].dtypes)
```

```python
df_clean["Net_Sales"] = (
    df_clean["Gross_Sales"] - df_clean["Discount_Amount"]
)

print("Net Sales calculated successfully!")
```

```python
df_clean[
    [
        "Product Type",
        "Unit Price",
        "Units Sold",
        "Discount (%)",
        "Gross_Sales",
        "Discount_Amount",
        "Net_Sales"
    ]
].head(10)
```

```python
total_gross_sales = df_clean["Gross_Sales"].sum()
total_discount = df_clean["Discount_Amount"].sum()
total_net_sales = df_clean["Net_Sales"].sum()
total_units = df_clean["Units Sold"].sum()

print("Total Gross Sales:", round(total_gross_sales, 2))
print("Total Discount:", round(total_discount, 2))
print("Total Net Sales:", round(total_net_sales, 2))
print("Total Units Sold:", total_units)
```
